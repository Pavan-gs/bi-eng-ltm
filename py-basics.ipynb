```python
def isprime(num):
    half = num//2
    for i in range(2,half+1):
        if num%i==0:
            print("the number is not a prime")
            break
    else:
        print("it is a prime number")
```

```python
isprime(23)
```

```python
for i in range(5):
    print(i, end='-->')
    isprime(i)
```
