```python
def new_decorator(actual_func):
    ''' This is a demo of a decorator function'''
    def wrapper_func():
        print("Statement before the execution of the function")
        actual_func()
        print("The statement after the execution of the function")
    return wrapper_func
```

```python
@new_decorator
def new_func():
    print("Hello I am yet to be decorated")
new_func()
```

```python
import numpy as np

class FitnessTrackerAnalyzer:

    def create_fitness_array(self, fitness_values: list) -> np.ndarray:
        return np.array(fitness_values, dtype=float)

    def validate_fitness_array(self, fitness_array: np.ndarray) -> bool:
        if fitness_array.size == 0:
            return False
        if np.any(fitness_array <= 0):
            return False
        return True

    def compute_fitness_statistics(self, fitness_array: np.ndarray) -> tuple:
        total = float(np.sum(fitness_array))
        average = round(float(np.mean(fitness_array)), 1)
        maximum = float(np.max(fitness_array))
        return (total, average, maximum)

    def filter_high_intensity(self, fitness_array: np.ndarray) -> np.ndarray:
        fitness_array = fitness_array.astype(float)
        fitness_array[fitness_array >= 50.0] *= 0.9
        return fitness_array

    def label_high_activity(self, fitness_array: np.ndarray) -> np.ndarray:
        mean_val = np.mean(fitness_array)
        return np.array(["High" if x > mean_val else "Normal" for x in fitness_array])

    def format_fitness_readings(self, fitness_array: np.ndarray) -> np.ndarray:
        return np.array([f"{x:.2f} bpm" for x in fitness_array])
```

```python
analyzer = FitnessTrackerAnalyzer()

values = [72, 85, 120, 145, 90]
arr = analyzer.create_fitness_array(values)

if analyzer.validate_fitness_array(arr):
    print(analyzer.compute_fitness_statistics(arr))
    print(analyzer.filter_high_intensity(arr.copy()))
    print(analyzer.label_high_activity(arr))
    print(analyzer.format_fitness_readings(arr))
```

```python
class book:
    def __init__(self,title,author,pages):
        self.title = title
        self.author = author
        self.pages = pages
    
    def __str__(self):
        return "The {} is written by {} and has {} pages" .format(self.title,self.author,self.pages)
    
    def __len__(self):
        return self.pages
    
    def __del__(self):
        print("The book is deleted")
```
